## Nivel 1

### 1. Consulta la API pública JSONPlaceholder utilizando el método GET para obtener:

**Lista de posts**

````python
import requests

url = 'https://jsonplaceholder.typicode.com/'
posts = requests.get(url + 'posts')
print('Code ' + str(posts.status_code))
print('Response: ')
print(posts.text)
````

**Lista de usuarios**

````python
usuarios = requests.get(url + 'users')
print('Code ' + str(usuarios.status_code))
print('Response: ')
print(usuarios.text)
````

**Lista de Todos**

````python
todos = requests.get(url + 'todos')
print('Code:', todos.status_code)
print('Response: ')
print(todos.text)
````

### 2. Muestra por pantalla:

- La cantidad total de cada recurso.
- El código de estado de cada petición.

````python
print('Posts: Code:', posts.status_code, '- Cantidad:', len(posts.json()))
print('Usuarios: Code:', usuarios.status_code, '- Cantidad:', len(usuarios.json()))
print('Todos: Code:', todos.status_code, '- Cantidad:', len(todos.json()))
````

### 3. Realiza una petición a una publicación inexistente para obtener un error 404 y muestra el código de estado recibido.

````python
response = requests.get(url + 'posts/1234567')
print('Code:', response.status_code)
````

### 4. Realiza una petición POST para crear una nueva publicación ficticia. Incluye el título, el cuerpo del mensaje y un userId.

- Muestra la respuesta JSON.
- Muestra el código de estado.

````python
post = {
    'title': 'post de prueba',
    'body': 'este es un post de prueba',
    'userId': 1,
}
response = requests.post(url + 'posts', json=post)
print('Code:', response.status_code)
print('Response:', response.text)
````

### 5. Realiza una petición PATCH para modificar parcialmente una publicación existente.

- Muestra la respuesta JSON.
- Muestra el código de estado.

````python
post = {
    'body': 'este es un post de prueba que ha sido modificado',
}
response = requests.patch(url + 'posts/101', json=post)
print('Code:', response.status_code)
print('Response:', response.text)
````

### 6. Haz una petición DELETE sobre una publicación.

Muestra la respuesta JSON.
Muestra el código de estado.

````python
response = requests.delete(url + 'posts/101')
print('Code:', response.status_code)
print('Response:', response.text)
````
