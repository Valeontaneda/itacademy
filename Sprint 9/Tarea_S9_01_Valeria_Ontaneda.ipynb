### **1. Calculadora del índice de masa corporal**

Escribe una función que calcule el IMC ingresado por el usuario/a, es decir, quien lo ejecute deberá ingresar estos datos. Puedes obtener más información de su cálculo en: 🔗 Índice de masa corporal IMC que es y cómo se calcula.

La función debe clasificar el resultado en sus respectivas categorías.

Consejo: Intenta validar los datos previamente, para que envíe un mensaje de advertencia si los datos introducidos por el usuario/a no están en el formato adecuado o no toma valores razonables.

```python
def measure_request(prompt, max_value):
    while True:
        try:
            # input and safegard for floats and values with commas
            value = float(input(prompt).replace(",","."))
            # limit value input using raise ValueError
            if value <= 0 or value > max_value:
                raise ValueError
        except ValueError:
            print("Make sure to enter a number and a reasonable measure")
        else:
            return value
        
def imc_categorization(weight, height):
    imc = weight / (height ** 2)
    if imc < 18.5:
        print("Your result: Underweight")
    elif imc < 25:
        print("Your result: Normal weight")
    elif imc < 30:
        print("Your result: Overweight")
    else: 
        print("Your result: Obesity")

user_weight = measure_request("Insert your weight in kg", 600)
user_height = measure_request("Insert your height in meters", 3)

imc_categorization(user_weight, user_height)
```

### Note: 
Using a single function to request measures (height and weight) and to add safeguards such as making sure the value that is returned is float and is a reasonable measurement (using limits to avoid negative numbers or numbers that are too high based on each measure)

```python
import pandas as pd

# defining temp values that are accepted 
valid_temps = ["celsius", "fahrenheit", "kelvin"]

# Creating a dataframe to store all conversion formulas as string
df = {
    "celsius": ["x", "(x - 32) * (5/9)", "x - 273.15"],
    "fahrenheit": ["x * 1.8 + 32", "x", "(x - 273.15) * 1.8 + 32"],
    "kelvin": ["x + 273.15", "(x - 32) * (5/9) + 273.15", "x"]
}

df = pd.DataFrame(df, index = ["celsius", "fahrenheit", "kelvin"])

def temp_request():
    while True:
        try:
            temp_start = input('Insert start temperature').lower()
            if temp_start not in valid_temps:
                raise ValueError  # added safeguard to only accept the valid temperatures using raise ValueError
            value = float(input('Insert value').replace(",","."))
        except ValueError:
            print(f"Make sure to enter a valid temperature {valid_temps} and a number as value")
        else:
            return temp_start, value
        
def temp_converter(temp_start, value, df = df):
    x = value # defining the value that the function receives as x which will be used for the calculations in the temp formulas
    results = {}
    for temp_final in valid_temps:
        formula = df.loc[temp_start, temp_final] 
        results[temp_final] = round(eval(formula), 2) # using eval to do the calculations in the strings
    return results


print(temp_converter(*temp_request()))
```

### Note
Using dataframes to keep the information of the formulas in strings and then evaluating the expressions with eval to get the results of each conversion.

### **2. Conversión de tipos de datos.**

El cliente recibe una lista de datos y necesita generar dos listas, la primera donde estarán todos los elementos que pudieron convertirse en flotantes y la otra donde están los elementos que no pudieron convertirse. 

```python
starter_list = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

def list_flatenning(input_list):
    flatten_list = []
    for item in input_list:
        if isinstance(item, (tuple, list, set)): # using isinstance to determine specific data types
            for subitem in item: # loop of subitem within original list
                flatten_list.append(subitem)
        else:
            flatten_list.append(item)
    return flatten_list

def float_convertor(input_list):
    float_list = []
    string_list = []
    for item in input_list:
        try:
            float_list.append(float(item))
        except ValueError:
            string_list.append(item)
    return float_list, string_list

float_list, string_list = float_convertor(list_flatenning(starter_list)) # calling functions in a way that it unpacks values returned

print("float_list:", float_list)
print("string_list:", string_list)
```

### **1. Generador de Contraseñas**

Explora el funcionamiento del módulo random de la librería numpy. 🔗 Random module in NumPy

En este punto, el cliente ha detectado un problema con las contraseñas que utilizan sus trabajadores. Asdf1234, fechas de cumpleaños o similares. Para solucionarlo, nos ha encargado una función de Python que genere contraseñas más seguras. La función debe depender de los siguientes parámetros:

- longitud (int): Longitud de la contraseña
- mayúsculas (bool = True): Si tiene que aparecer mayúsculas
- minúsculas (bool = True): Si tiene que aparecer minúsculas
- números (bool = True): Si tiene que aparecer números
- signos (bool = False): Si tiene que aparecer caracteres especiales (,-$? o similares)

Así pues, si ejecutamos la función de la siguiente forma:

crear_contraseña(10, True, True, True, True)

Deberíamos obtener un output (que debemos poder guardar) del estilo:

9Er,5Vn8P$

Asegúrate de que se cumplan todos los criterios, y que estas contraseñas sean realmente aleatorias.

(EXTRA) Explora cómo podríamos hacer que la función copiara la contraseña automáticamente en el portapapeles del ordenador (como si la hubiéramos seleccionado y hecho ctrl+copy).

```python
import numpy as np
import string # using string to get collection of different characters that will be used for the password generation
import pyperclip # importing pyperclip module to copy password in clipboard


def generate_pass(length, lower = True, upper = True, number = True, punctuation = True):
    
    if length >= 8: # validation of minimum length of 8 characters

        password = []
        total_list = []

        # using np.random.permutation to shuffle characters and string to get lowercase, uppercase, digits and punctuation collections
        np_lower_case = np.random.permutation(list(string.ascii_lowercase))
        np_upper_case = np.random.permutation(list(string.ascii_uppercase))
        np_number = np.random.permutation(list(string.digits))
        np_symbol = np.random.permutation(list(string.punctuation))
        
        if lower:
            password.append(str(np_lower_case[0]))
            total_list += list(string.ascii_lowercase)
        if upper:
            password.append(str(np_upper_case[0]))
            total_list += list(string.ascii_uppercase)
        if number:
            password.append(str(np_number[0]))
            total_list += list(string.digits)
        if punctuation:
            password.append(str(np_symbol[0]))
            total_list += list(string.punctuation)

        if len(password) > 1: # adding safeguard to ensure user picks at least 2 character types. Checking len password as the base has the first characters added
            filler_len = length - len(password) # based on the len argument passed in the function, calculate how many characters are missing to match the length
            np_filler = np.random.choice(total_list, size=filler_len) # using random.choice to add missing characters in a randomize way

            np_password = np.concatenate((np.array(password), np_filler)) # using concatenate to merge the base password with the filler one
            np.random.shuffle(np_password) # using shuffle to randomize the final array and make sure there are no patterns

            string_password = "".join(np_password) # using join to change array to string

            pyperclip.copy(string_password) # using pyperclip to copy pass to the clipboard
            print("The password has been successfully copied to your clipboard")
            
            return string_password
        
        else:
            print("You must select at least two type of characters")
    else:
        print("Your password must be at least 8 characters long")

generate_pass(8, True, True, True, True)
```

### Note
- Using numpy, string and pyperclip for this password generation.
- There are two safeguards: a min of 8 characters in the password and checking that at least 2 character types have been requested
- Using string to get collections for each character type and permutation from numpy to shuffle the first arrays as the idea is to get the first value [0] to add to the password to make sure at least one character of the types requested is in the password.
- Checking type of characters required, adding the first character of each to the password and creating a list with a mix of type of characters which then will serve to fill the rest of the password.
- Calculating how many characters are left to fill based on the length passed by user and the characters already added to the password
- Using np.choice to get characters from the list with all characters so that it is randomize
- Concatenating both arrays and shuffling again before changing it to a string using join() 
- Finally using pyperclip to copy password to the clipboard


### **2. Procesamiento de datos simple**

Una compañera de trabajo nos ha pedido un favor, aprovechando que sabe que estamos aprendiendo a programar. Tiene un histórico de partidos de fútbol catalán en un fichero, donde se almacenan los nombres de los equipos y los resultados. Necesita que procesemos los datos de forma automática, para extraer los resultados que necesita.

Utiliza el archivo "historic_partits.txt"

Necesita un programa que devuelva:

- El número total de goles que ha marcado cada equipo.
- El nombre del equipo más goleador
- El nombre del equipo más goleado
- La clasificación global (cada victoria: 3 pts, empate 1 pts, derrota 0 pts)


```python
import pandas as pd

# importing data with pandas using tab separator and column names
matches_df = pd.read_csv("/Users/valeriaontaneda/Documents/It_Academy/Especializacion/Sprint 9/historic partits.txt", sep = "\t|-", names = ["home_team", "score_home_team", "score_away_team", "away_team"], engine='python')
matches_df.info() # checking info imported with info and head
matches_df.head()
```

Using a function to add four new columns to understand the results and scores of the home and away teams 
Using apply to use function in the dataframe created and with axis = 1 to go row by row.
The lambda function receives data of each row as series and from each series passes the arguments needed for the function to return the tuple with the results that is then expanded and creates the 4 new columns with info from each row.

```python
def points_calculation(score_home_team, score_away_team):
    if score_home_team > score_away_team:
        return "win", 3, "lost", 0
    elif score_home_team < score_away_team:
        return "lost", 0, "win", 3
    else:
        return "tie", 1, "tie", 1

# adding 4 new columns and passing the score columns as arguments using apply
matches_df[['result_home_team', 'points_home_team', 'result_away_team', 'points_away_team']] = matches_df[['score_home_team', 'score_away_team']].apply(
    lambda row: points_calculation(row['score_home_team'], row['score_away_team']), axis=1, result_type='expand'
) #using axis=1 to go row by row instead of column by column and expand to separate the tuple in different columns

matches_df.head(10)
```

```python
home_matches_df = matches_df[["home_team","score_home_team", "score_away_team", "result_home_team", "points_home_team"]]
home_matches_df.rename(columns={"home_team" : "teams", "score_home_team" : "goals", "score_away_team" : "goals_received", "result_home_team" : "result", "points_home_team" : "points"}, inplace=True)

away_matches_df = matches_df[["away_team","score_away_team", "score_home_team", "result_away_team", "points_away_team"]]
away_matches_df.rename(columns={"away_team" : "teams", "score_away_team" : "goals", "score_home_team" : "goals_received", "result_away_team" : "result", "points_away_team" : "points"}, inplace=True)

final_matches_df = pd.concat([home_matches_df, away_matches_df], ignore_index=True) # using concat to unify home and away teams with its information
final_matches_df.head()
```

```python
final_matches_df['win'] = (final_matches_df['result'] == 'win').astype(int) # creating columns where win is true and changes to numeric value with astype(int)
final_matches_df['tie'] = (final_matches_df['result'] == 'tie').astype(int)
final_matches_df['lost'] = (final_matches_df['result'] == 'lost').astype(int)

score_board_df = final_matches_df.groupby("teams")[['win', 'tie', 'lost', 'goals', 'goals_received', 'points']].sum().sort_values("points", ascending=False)

score_board_df.head(20)
```

```python
team_goals = score_board_df[["goals"]].sort_values("goals", ascending=False)
team_goals.head(20)
```

```python
print(team_goals[team_goals.goals == team_goals.goals.max()]) # filter df to show the teams that match the max goals which works to show ties
```

```python
team_goals_received = score_board_df[["goals_received"]].sort_values("goals_received", ascending=False)
print(team_goals_received[team_goals_received.goals_received == team_goals_received.goals_received.max()])
```
